```python
# %% Chargement direct de 3M lignes (SANS fichier intermédiaire)
import pandas as pd
import numpy as np
import time
import gc

print("="*70)
print(" CHARGEMENT DIRECT DE 3M LIGNES (OPTIMISÉ RAM)")
print("="*70)

TARGET_SIZE = 3_000_000
print(f" Objectif : {TARGET_SIZE:,} lignes\n")

# Récupérer les fichiers CSV
csv_files = sorted(Config.DATA_DIR.glob("*.csv"))
csv_files = [f for f in csv_files if 'sample' not in f.name.lower()]

# Comptage rapide
file_sizes = [
    (csv_files[0], 1048575),
    (csv_files[1], 1048575),
    (csv_files[2], 1048575),
    (csv_files[3], 7948748),
    (csv_files[4], 1048575),
    (csv_files[5], 1048575),
    (csv_files[6], 1048575),
    (csv_files[7], 613104),
    (csv_files[8], 331125),
    (csv_files[9], 1048575),
]
total_lines = sum(n for _, n in file_sizes)

print(" Chargement progressif avec limite 3M...\n")

start_time = time.time()
dfs = []
lines_loaded = 0

for i, (csv_file, n_lines) in enumerate(file_sizes, 1):
    # Calculer combien prendre de ce fichier
    remaining = TARGET_SIZE - lines_loaded
    if remaining <= 0:
        break

    n_to_load = min(remaining, int((n_lines / total_lines) * TARGET_SIZE * 1.3))

    print(f"   [{i}/{len(file_sizes)}] {csv_file.name[:40]:<40}", end=" ")

    # Charger avec échantillonnage
    if n_to_load < n_lines:
        np.random.seed(42 + i)
        skip_indices = np.random.choice(
            range(1, n_lines + 1),
            size=n_lines - n_to_load,
            replace=False
        )
        skip_set = set(skip_indices)

        df_chunk = pd.read_csv(
            csv_file,
            skiprows=lambda x: x in skip_set,
            low_memory=False,
            on_bad_lines='skip'
        )
    else:
        df_chunk = pd.read_csv(
            csv_file,
            low_memory=False,
            on_bad_lines='skip'
        )

    lines_loaded += len(df_chunk)
    dfs.append(df_chunk)

    print(f"→ +{len(df_chunk):,} | Total: {lines_loaded:,}/{TARGET_SIZE:,}")

    # ARRÊTER si on a atteint l'objectif
    if lines_loaded >= TARGET_SIZE:
        print(f"\n    Objectif atteint ! Arrêt du chargement.")
        break

# Concaténer
print(f"\n🔗 Concaténation de {len(dfs)} DataFrames...")
df = pd.concat(dfs, ignore_index=True)

# Libérer
del dfs
gc.collect()

print(f" Chargé en {time.time()-start_time:.1f}s")

# Réduire à EXACTEMENT 3M si nécessaire
if len(df) > TARGET_SIZE:
    print(f"\n Réduction à exactement {TARGET_SIZE:,} lignes...")
    df = df.sample(n=TARGET_SIZE, random_state=42).reset_index(drop=True)

print(f"\n Résultat final :")
print(f"   Shape : {df.shape}")
print(f"   Lignes : {df.shape[0]:,}")
print(f"   Colonnes : {df.shape[1]}")
print(f"   Mémoire : {df.memory_usage(deep=True).sum()/(1024**2):.1f} MB")

# Distribution des classes
if 'Label' in df.columns:
    print(f"\n Distribution des classes :")
    label_counts = df['Label'].value_counts()
    for label, count in label_counts.head(10).items():
        pct = count / len(df) * 100
        print(f"   {str(label)[:40]:<40s} : {count:>10,} ({pct:>5.2f}%)")
    if len(label_counts) > 10:
        print(f"   ... et {len(label_counts)-10} autres classes")

# Sauvegarder
sample_path = Config.DATA_DIR / "cicids2018_final_3M.csv"
print(f"\n Sauvegarde...")
df.to_csv(sample_path, index=False)
print(f"    Sauvegardé : {sample_path.name}")

print("\n" + "="*70)
print(" DATASET DE 3M PRÊT POUR LES GNN")
print("="*70)
```

```python
# %% CORRECTION : Rechargement avec typage correct
import pandas as pd
import numpy as np
from sklearn.preprocessing import LabelEncoder, StandardScaler
from sklearn.model_selection import train_test_split
import time
import gc

print("="*70)
print("🔧 PRÉTRAITEMENT CORRIGÉ (TYPAGE + NETTOYAGE)")
print("="*70)

start_time = time.time()

# ============================================================
# 1. RECHARGER LE FICHIER AVEC FORÇAGE DU TYPAGE
# ============================================================
print(f"\n Rechargement du fichier avec typage forcé...")

sample_path = Config.DATA_DIR / "cicids2018_final_3M.csv"
df = pd.read_csv(sample_path, on_bad_lines='skip')

print(f"    Chargé : {df.shape}")

# ============================================================
# 2. IDENTIFIER LA COLONNE LABEL
# ============================================================
target_col = 'Label'
y = df[target_col].copy()
X = df.drop(columns=[target_col])

print(f"\n Dataset initial :")
print(f"   Features (X) : {X.shape}")
print(f"   Target (y) : {len(y)}")
print(f"   Classes : {y.nunique()}")

# ============================================================
# 3. NETTOYER LES COLONNES NON-NUMÉRIQUES (VRAI NETTOYAGE)
# ============================================================
# ============================================================
# 3. FEATURE ENGINEERING: Timestamp, Dst Port, Protocol
# ============================================================
print(f"\n🔧 Feature Engineering...")

# ────────────────────────────────────────────────────────────
# A. TIMESTAMP → Features Temporelles
# ────────────────────────────────────────────────────────────
if 'Timestamp' in X.columns:
    print(f"   Processing Timestamp...")
    try:
        # Convertir en datetime
        timestamps = pd.to_datetime(X['Timestamp'], errors='coerce')

        # Extraire features temporelles
        X['hour'] = timestamps.dt.hour.fillna(0).astype(int)
        X['day_of_week'] = timestamps.dt.dayofweek.fillna(0).astype(int)
        X['is_weekend'] = (timestamps.dt.dayofweek >= 5).fillna(False).astype(int)

        # Période de la journée
        X['time_period'] = pd.cut(
            timestamps.dt.hour.fillna(0),
            bins=[0, 6, 12, 18, 24],
            labels=[0, 1, 2, 3],  # Nuit, Matin, Après-midi, Soir
            include_lowest=True
        ).astype(int)

        # Supprimer Timestamp original
        X = X.drop(columns=['Timestamp'])
        print(f"      ✓ Features créées: hour, day_of_week, is_weekend, time_period")

    except Exception as e:
        print(f"      ⚠️  Erreur Timestamp: {e}")
        X = X.drop(columns=['Timestamp'], errors='ignore')

# ────────────────────────────────────────────────────────────
# B. DST PORT → Catégorisation
# ────────────────────────────────────────────────────────────
if 'Dst Port' in X.columns:
    print(f"   Processing Dst Port...")
    try:
        dst_port = pd.to_numeric(X['Dst Port'], errors='coerce').fillna(0).astype(int)

        # Catégories de ports
        X['port_is_well_known'] = (dst_port <= 1023).astype(int)  # Ports système
        X['port_is_registered'] = ((dst_port >= 1024) & (dst_port <= 49151)).astype(int)
        X['port_is_ephemeral'] = (dst_port >= 49152).astype(int)  # Ports dynamiques

        # Ports spécifiques critiques
        X['port_is_http'] = dst_port.isin([80, 8080, 443]).astype(int)
        X['port_is_ssh'] = (dst_port == 22).astype(int)
        X['port_is_rdp'] = (dst_port == 3389).astype(int)
        X['port_is_dns'] = (dst_port == 53).astype(int)
        X['port_is_ftp'] = dst_port.isin([20, 21]).astype(int)

        # Garder aussi le port normalisé
        X['dst_port_normalized'] = dst_port / 65535.0  # Normaliser 0-1

        # Supprimer Dst Port original
        X = X.drop(columns=['Dst Port'])
        print(f"      ✓ Features créées: 9 features de port")

    except Exception as e:
        print(f"      ⚠️  Erreur Dst Port: {e}")
        X = X.drop(columns=['Dst Port'], errors='ignore')

# ────────────────────────────────────────────────────────────
# C. PROTOCOL → Encodage
# ────────────────────────────────────────────────────────────
if 'Protocol' in X.columns:
    print(f"   Processing Protocol...")
    try:
        # Convertir protocol en numérique si nécessaire
        protocol = X['Protocol']

        # Si c'est du texte (TCP, UDP, ICMP), mapper vers codes
        if protocol.dtype == 'object':
            protocol_map = {
                'TCP': 6,
                'UDP': 17,
                'ICMP': 1,
                'IGMP': 2,
                'IPv6': 41,
                'RSVP': 46,
                'GRE': 47
            }
            protocol = protocol.map(protocol_map).fillna(0).astype(int)
        else:
            protocol = pd.to_numeric(protocol, errors='coerce').fillna(0).astype(int)

        # Créer features binaires pour principaux protocoles
        X['protocol_is_tcp'] = (protocol == 6).astype(int)
        X['protocol_is_udp'] = (protocol == 17).astype(int)
        X['protocol_is_icmp'] = (protocol == 1).astype(int)

        # Garder aussi la valeur numérique normalisée
        X['protocol_value'] = protocol / 255.0  # Normaliser

        # Supprimer Protocol original
        X = X.drop(columns=['Protocol'])
        print(f"      ✓ Features créées: 4 features de protocole")

    except Exception as e:
        print(f"      Erreur Protocol: {e}")
        X = X.drop(columns=['Protocol'], errors='ignore')

print(f"   ✓ Feature engineering terminé")
print(f"   Features totales maintenant: {X.shape[1]}")

# Convertir TOUT en numérique (forcer)
print(f"    Conversion forcée en numérique...")
for col in X.columns:
    X[col] = pd.to_numeric(X[col], errors='coerce')

# Supprimer les colonnes qui sont devenues 100% NaN
all_nan_cols = X.columns[X.isnull().all()].tolist()
if all_nan_cols:
    print(f"   Suppression colonnes 100% NaN : {len(all_nan_cols)}")
    X = X.drop(columns=all_nan_cols)

# Supprimer colonnes constantes
constant_cols = X.columns[X.nunique() <= 1].tolist()
if constant_cols:
    print(f"   Suppression colonnes constantes : {len(constant_cols)}")
    X = X.drop(columns=constant_cols)

print(f"    Features finales : {X.shape[1]}")

# ============================================================
# 4. NETTOYER LES CLASSES (SUPPRIMER CLASSES AVEC <10 SAMPLES)
# ============================================================
print(f"\n  Nettoyage des classes...")

label_counts = y.value_counts()
print(f"   Classes avant : {len(label_counts)}")

# Identifier les classes à supprimer
min_samples = 100  # Minimum pour GNN
classes_to_remove = label_counts[label_counts < min_samples].index.tolist()

if classes_to_remove:
    print(f"     Suppression de {len(classes_to_remove)} classes < {min_samples} échantillons :")
    for cls in classes_to_remove:
        count = label_counts[cls]
        print(f"      - {cls[:40]:<40s} : {count}")

    # Supprimer ces lignes
    mask = ~y.isin(classes_to_remove)
    y = y[mask]
    X = X[mask]

    print(f"    Lignes restantes : {len(X):,}")
    print(f"    Classes restantes : {y.nunique()}")

# ============================================================
# 5. VALEURS MANQUANTES
# ============================================================
print(f"\n Valeurs manquantes...")
n_missing = X.isnull().sum().sum()
print(f"   Total : {n_missing:,}")

if n_missing > 0:
    print(f"    Remplacement par la médiane...")
    X = X.fillna(X.median(numeric_only=True))

    # Vérifier s'il reste des NaN
    remaining = X.isnull().sum().sum()
    if remaining > 0:
        print(f"     {remaining} NaN restants → remplacement par 0")
        X = X.fillna(0)

# ============================================================
# 6. VALEURS INFINIES
# ============================================================
print(f"\n∞ Valeurs infinies...")
inf_mask = np.isinf(X.values)
n_inf = inf_mask.sum()

if n_inf > 0:
    print(f"   Total : {n_inf:,}")
    X = X.replace([np.inf, -np.inf], np.nan)
    X = X.fillna(X.median(numeric_only=True))
    X = X.fillna(0)
    print(f"    Corrigées")

# ============================================================
# 7. ENCODAGE DES LABELS
# ============================================================
print(f"\n  Encodage des labels...")

le = LabelEncoder()
y_encoded = le.fit_transform(y)

print(f"   Classes : {len(le.classes_)}")
for i, class_name in enumerate(le.classes_):
    count = (y_encoded == i).sum()
    pct = count / len(y_encoded) * 100
    print(f"      {i:2d} → {class_name[:35]:<35s} : {count:>10,} ({pct:>5.2f}%)")

# ============================================================
# 8. SPLIT TRAIN/TEST
# ============================================================
print(f"\n  Split train/test (70/30)...")

X_train, X_test, y_train, y_test = train_test_split(
    X.values,  # Convertir en array
    y_encoded,
    test_size=0.3,
    random_state=42,
    stratify=y_encoded
)

print(f"   Train : {X_train.shape[0]:,} échantillons")
print(f"   Test  : {X_test.shape[0]:,} échantillons")

# Libérer
del X, y, df
gc.collect()

# ============================================================
# 9. NORMALISATION
# ============================================================
print(f"\n Normalisation...")

scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

print(f"    Mean : {X_train_scaled.mean():.4f}")
print(f"    Std  : {X_train_scaled.std():.4f}")

del X_train, X_test
gc.collect()

# ============================================================
# 10. RÉSUMÉ
# ============================================================
elapsed = time.time() - start_time

print(f"\n" + "="*70)
print(f" PRÉTRAITEMENT TERMINÉ EN {elapsed:.1f}s")
print(f"="*70)

print(f"\n Données finales :")
print(f"   X_train_scaled : {X_train_scaled.shape}")
print(f"   X_test_scaled  : {X_test_scaled.shape}")
print(f"   Features       : {X_train_scaled.shape[1]}")
print(f"   Classes        : {len(le.classes_)}")

print(f"\n Mémoire :")
print(f"   Train : {X_train_scaled.nbytes/(1024**2):.1f} MB")
print(f"   Test  : {X_test_scaled.nbytes/(1024**2):.1f} MB")

# Métadonnées
class_names = le.classes_
num_features = X_train_scaled.shape[1]
num_classes = len(class_names)

print(f"\n PRÊT POUR LES GNN !")
print(f"   {num_features} features → {num_classes} classes")
print("="*70)
```

```python
# %% Construction du graphe k-NN (OPTIMISÉ pour 100K nœuds)
from sklearn.neighbors import NearestNeighbors
import networkx as nx
import time
import gc

print("="*70)
print("🔨 CONSTRUCTION DU GRAPHE k-NN")
print("="*70)

start_time = time.time()

# ============================================================
# PARAMÈTRES DU GRAPHE
# ============================================================
K_NEIGHBORS = 10  # Nombre de voisins (10-15 recommandé)

print(f"\n⚙️  Paramètres :")
print(f"   Nœuds : {X_combined.shape[0]:,}")
print(f"   Features : {X_combined.shape[1]}")
print(f"   k (voisins) : {K_NEIGHBORS}")
print(f"   Métrique : euclidean")

# ============================================================
# CALCUL DES k PLUS PROCHES VOISINS
# ============================================================
print(f"\n Calcul des {K_NEIGHBORS} plus proches voisins...")
print(f"   (Cette étape peut prendre 2-5 minutes)")

# NearestNeighbors avec ball_tree (optimisé pour 100K)
nbrs = NearestNeighbors(
    n_neighbors=K_NEIGHBORS + 1,  # +1 car le nœud lui-même est inclus
    algorithm='ball_tree',  # Plus rapide pour grands datasets
    metric='euclidean',
    n_jobs=-1  # Utiliser tous les CPU
)

nbrs.fit(X_combined)
distances, indices = nbrs.kneighbors(X_combined)

elapsed_knn = time.time() - start_time
print(f"    k-NN calculé en {elapsed_knn:.1f}s")

# ============================================================
# CONSTRUCTION DU GRAPHE NETWORKX
# ============================================================
print(f"\n Construction du graphe NetworkX...")

# Créer la liste d'arêtes
edges = []
for i in range(len(X_combined)):
    for j_idx in range(1, K_NEIGHBORS + 1):  # Skip index 0 (self)
        j = indices[i, j_idx]
        if i < j:  # Éviter les doublons (graphe non-dirigé)
            edges.append((i, j))

print(f"   Arêtes brutes : {len(edges):,}")

# Créer le graphe
G = nx.Graph()
G.add_nodes_from(range(len(X_combined)))
G.add_edges_from(edges)

elapsed_graph = time.time() - start_time
print(f"    Graphe créé en {elapsed_graph:.1f}s")

# ============================================================
# STATISTIQUES DU GRAPHE
# ============================================================
print(f"\n📊 Statistiques du graphe :")
print(f"   Nœuds : {G.number_of_nodes():,}")
print(f"   Arêtes : {G.number_of_edges():,}")
print(f"   Densité : {nx.density(G):.6f}")
print(f"   Degré moyen : {sum(dict(G.degree()).values()) / G.number_of_nodes():.2f}")

# Degré min/max
degrees = dict(G.degree())
print(f"   Degré min : {min(degrees.values())}")
print(f"   Degré max : {max(degrees.values())}")

# Connexité
is_connected = nx.is_connected(G)
print(f"   Connexe : {'Oui' if is_connected else 'Non'}")

if not is_connected:
    components = list(nx.connected_components(G))
    print(f"   Composantes : {len(components)}")
    print(f"   Plus grande : {len(max(components, key=len)):,} nœuds")

# Libérer mémoire
del edges, distances, indices, nbrs
gc.collect()

print(f"\n⏱️  Temps total construction : {time.time() - start_time:.1f}s")

print("\n" + "="*70)
print("GRAPHE k-NN CONSTRUIT AVEC SUCCÈS")
print("="*70)
```

```python
# %% Définition des 4 modèles GNN classiques
import torch
import torch.nn.functional as F
from torch_geometric.nn import GCNConv, GATConv, SAGEConv, InnerProductDecoder

print("="*70)
print(" DÉFINITION DES 4 MODÈLES GNN CLASSIQUES")
print("="*70)

# ============================================================
# 1. GCN (Graph Convolutional Network)
# ============================================================
class GCN(torch.nn.Module):
    """
    Graph Convolutional Network (Kipf & Welling, 2017)
    Modèle de base pour apprentissage transductif.
    """
    def __init__(self, num_features, num_classes, hidden_dim=64, dropout=0.5):
        super(GCN, self).__init__()
        self.conv1 = GCNConv(num_features, hidden_dim)
        self.conv2 = GCNConv(hidden_dim, num_classes)
        self.dropout = dropout

    def forward(self, data):
        x, edge_index = data.x, data.edge_index

        # Couche 1 : GCN + ReLU + Dropout
        x = self.conv1(x, edge_index)
        x = F.relu(x)
        x = F.dropout(x, p=self.dropout, training=self.training)

        # Couche 2 : GCN
        x = self.conv2(x, edge_index)

        return F.log_softmax(x, dim=1)

print(" GCN défini")

# ============================================================
# 2. GAT (Graph Attention Network)
# ============================================================
class GAT(torch.nn.Module):
    """
    Graph Attention Network (Veličković et al., 2018)
    Utilise un mécanisme d'attention multi-têtes.
    """
    def __init__(self, num_features, num_classes, hidden_dim=64, heads=8, dropout=0.6):
        super(GAT, self).__init__()

        # Couche 1 : Multi-head attention
        self.conv1 = GATConv(num_features, hidden_dim, heads=heads, dropout=dropout)

        # Couche 2 : Single-head attention
        self.conv2 = GATConv(hidden_dim * heads, num_classes, heads=1,
                            concat=False, dropout=dropout)
        self.dropout = dropout

    def forward(self, data):
        x, edge_index = data.x, data.edge_index

        # Dropout sur features d'entrée
        x = F.dropout(x, p=self.dropout, training=self.training)

        # Couche 1 : GAT + ELU
        x = self.conv1(x, edge_index)
        x = F.elu(x)

        # Dropout
        x = F.dropout(x, p=self.dropout, training=self.training)

        # Couche 2 : GAT
        x = self.conv2(x, edge_index)

        return F.log_softmax(x, dim=1)

print(" GAT défini")

# ============================================================
# 3. GraphSAGE
# ============================================================
class GraphSAGE(torch.nn.Module):
    """
    GraphSAGE (Hamilton et al., 2017)
    Échantillonnage et agrégation du voisinage.
    Adapté pour apprentissage inductif.
    """
    def __init__(self, num_features, num_classes, hidden_dim=64, dropout=0.5):
        super(GraphSAGE, self).__init__()
        self.conv1 = SAGEConv(num_features, hidden_dim)
        self.conv2 = SAGEConv(hidden_dim, num_classes)
        self.dropout = dropout

    def forward(self, data):
        x, edge_index = data.x, data.edge_index

        # Couche 1 : SAGE + ReLU + Dropout
        x = self.conv1(x, edge_index)
        x = F.relu(x)
        x = F.dropout(x, p=self.dropout, training=self.training)

        # Couche 2 : SAGE
        x = self.conv2(x, edge_index)

        return F.log_softmax(x, dim=1)

print(" GraphSAGE défini")

# ============================================================
# 4. GAE (Graph Auto-Encoder) - ENCODEUR SEULEMENT
# ============================================================
class GCNEncoder(torch.nn.Module):
    """
    Encodeur GCN pour GAE.
    Encode les nœuds dans un espace latent.
    """
    def __init__(self, num_features, hidden_dim=32, latent_dim=16):
        super(GCNEncoder, self).__init__()
        self.conv1 = GCNConv(num_features, hidden_dim)
        self.conv2 = GCNConv(hidden_dim, latent_dim)

    def forward(self, x, edge_index):
        x = self.conv1(x, edge_index)
        x = F.relu(x)
        x = self.conv2(x, edge_index)
        return x



# ============================================================
# 5. RÉSUMÉ DES MODÈLES
# ============================================================
print(f"\n" + "="*70)
print(f" RÉSUMÉ DES MODÈLES")
print(f"="*70)

print(f"\n GCN (Graph Convolutional Network)")
print(f"   • Couches : 2 GCN")
print(f"   • Dropout : 0.5")
print(f"   • Adapté pour : Transductif")

print(f"\n GAT (Graph Attention Network)")
print(f"   • Couches : 2 GAT (multi-head puis single-head)")
print(f"   • Heads : 8 (couche 1)")
print(f"   • Dropout : 0.6")
print(f"   • Adapté pour : Transductif")

print(f"\n GraphSAGE")
print(f"   • Couches : 2 SAGE")
print(f"   • Dropout : 0.5")
print(f"   • Adapté pour : Inductif ET Transductif")



# ============================================================
# 6. TEST DE CRÉATION DES MODÈLES
# ============================================================
print(f"\n" + "="*70)
print(f" TEST DE CRÉATION DES MODÈLES")
print(f"="*70)

num_features = data.num_node_features
num_classes = data.y.max().item() + 1

print(f"\n  Paramètres :")
print(f"   • Features : {num_features}")
print(f"   • Classes : {num_classes}")

# Créer les modèles
device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
print(f"   • Device : {device}")

print(f"\n Création des modèles...")

model_gcn = GCN(num_features, num_classes, hidden_dim=64, dropout=0.5).to(device)
model_gat = GAT(num_features, num_classes, hidden_dim=8, heads=8, dropout=0.6).to(device)
model_sage = GraphSAGE(num_features, num_classes, hidden_dim=64, dropout=0.5).to(device)

# Compter les paramètres
def count_parameters(model):
    return sum(p.numel() for p in model.parameters() if p.requires_grad)

print(f"\n Nombre de paramètres :")
print(f"   • GCN : {count_parameters(model_gcn):,}")
print(f"   • GAT : {count_parameters(model_gat):,}")
print(f"   • GraphSAGE : {count_parameters(model_sage):,}")

# Test forward pass
print(f"\n Test forward pass (GCN)...")
data_test = data.to(device)
model_gcn.eval()
with torch.no_grad():
    out = model_gcn(data_test)
print(f"    Output shape : {out.shape}")
print(f"    Output type : {out.dtype}")

# Libérer
data_test = data_test.to('cpu')
torch.cuda.empty_cache() if torch.cuda.is_available() else None

print(f"\n" + "="*70)
print(f" TOUS LES MODÈLES CRÉÉS ET TESTÉS AVEC SUCCÈS")
print(f"="*70)
```

```python
# %% Entraînement MODE TRANSDUCTIF (GCN, GAT, GraphSAGE)
import torch
import torch.nn.functional as F
import time
from sklearn.metrics import (
    accuracy_score, f1_score, precision_score, recall_score,
    matthews_corrcoef, cohen_kappa_score, balanced_accuracy_score
)

print("="*70)
print(" ENTRAÎNEMENT MODE TRANSDUCTIF")
print("="*70)

# ============================================================
# FONCTIONS D'ENTRAÎNEMENT ET D'ÉVALUATION
# ============================================================
def train_gnn(model, data, optimizer, device):
    """Entraîne le modèle pour une époque."""
    model.train()
    data = data.to(device)
    optimizer.zero_grad()

    out = model(data)
    loss = F.nll_loss(out[data.train_mask], data.y[data.train_mask])
    loss.backward()
    optimizer.step()

    return loss.item()

@torch.no_grad()
def evaluate_gnn(model, data, mask, device):
    """Évalue le modèle sur un split donné."""
    model.eval()
    data = data.to(device)

    out = model(data)
    pred = out.argmax(dim=1)

    # Accuracy
    correct = pred[mask] == data.y[mask]
    acc = int(correct.sum()) / int(mask.sum())

    # Loss
    loss = F.nll_loss(out[mask], data.y[mask]).item()

    # Prédictions pour métriques détaillées
    y_true = data.y[mask].cpu().numpy()
    y_pred = pred[mask].cpu().numpy()

    return acc, loss, y_true, y_pred

def compute_all_metrics(y_true, y_pred):
    """Calcule toutes les métriques (identiques aux modèles tabulaires)."""
    return {
        'Accuracy': accuracy_score(y_true, y_pred),
        'F1_weighted': f1_score(y_true, y_pred, average='weighted', zero_division=0),
        'F1_macro': f1_score(y_true, y_pred, average='macro', zero_division=0),
        'Precision_weighted': precision_score(y_true, y_pred, average='weighted', zero_division=0),
        'Recall_weighted': recall_score(y_true, y_pred, average='weighted', zero_division=0),
        'MCC': matthews_corrcoef(y_true, y_pred),
        'Cohen_Kappa': cohen_kappa_score(y_true, y_pred),
        'Balanced_Accuracy': balanced_accuracy_score(y_true, y_pred)
    }

# ============================================================
# CONFIGURATION
# ============================================================
device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
data = data.to(device)

EPOCHS = 200
PATIENCE = 30  # Early stopping

print(f"\n  Configuration :")
print(f"   • Device : {device}")
print(f"   • Époques max : {EPOCHS}")
print(f"   • Early stopping : {PATIENCE} époques")
print(f"   • Mode : TRANSDUCTIF (tous les nœuds visibles)")

# ============================================================
# ENTRAÎNEMENT DES 3 MODÈLES
# ============================================================
models_to_train = {
    'GCN': GCN(num_features, num_classes, hidden_dim=64, dropout=0.5),
    'GAT': GAT(num_features, num_classes, hidden_dim=8, heads=8, dropout=0.6),
    'GraphSAGE': GraphSAGE(num_features, num_classes, hidden_dim=64, dropout=0.5)
}

results_transductive = {}

print(f"\n" + "="*70)

for model_name, model in models_to_train.items():
    print(f"\n Entraînement {model_name}...")
    print(f"-" * 70)

    model = model.to(device)
    optimizer = torch.optim.Adam(model.parameters(), lr=0.01, weight_decay=5e-4)

    # Early stopping
    best_val_acc = 0
    best_val_loss = float('inf')
    patience_counter = 0
    best_model_state = None

    # Historique
    history = {
        'train_loss': [],
        'val_loss': [],
        'val_acc': []
    }

    start_time = time.time()

    # Entraînement
    for epoch in range(1, EPOCHS + 1):
        # Train
        train_loss = train_gnn(model, data, optimizer, device)

        # Evaluate
        train_acc, _, _, _ = evaluate_gnn(model, data, data.train_mask, device)
        val_acc, val_loss, _, _ = evaluate_gnn(model, data, data.val_mask, device)

        # Historique
        history['train_loss'].append(train_loss)
        history['val_loss'].append(val_loss)
        history['val_acc'].append(val_acc)

        # Early stopping
        if val_acc > best_val_acc:
            best_val_acc = val_acc
            best_val_loss = val_loss
            patience_counter = 0
            best_model_state = model.state_dict().copy()
        else:
            patience_counter += 1

        # Log
        if epoch % 20 == 0 or epoch == 1:
            print(f"   Epoch {epoch:3d} | Train Loss: {train_loss:.4f} | "
                  f"Val Acc: {val_acc:.4f} | Best Val: {best_val_acc:.4f}")

        # Stop si patience dépassée
        if patience_counter >= PATIENCE:
            print(f"     Early stopping à l'époque {epoch}")
            break

    # Charger le meilleur modèle
    if best_model_state is not None:
        model.load_state_dict(best_model_state)

    training_time = time.time() - start_time

    # Évaluation finale sur TEST
    print(f"\n    Évaluation finale sur le test set...")
    test_acc, test_loss, y_true_test, y_pred_test = evaluate_gnn(
        model, data, data.test_mask, device
    )

    # Métriques complètes
    metrics = compute_all_metrics(y_true_test, y_pred_test)
    metrics['Model'] = model_name
    metrics['Training_Time'] = training_time
    metrics['Best_Val_Acc'] = best_val_acc
    metrics['Test_Acc'] = test_acc
    metrics['Mode'] = 'Transductif'

    # Sauvegarder
    results_transductive[model_name] = {
        'model': model,
        'metrics': metrics,
        'history': history,
        'predictions': (y_true_test, y_pred_test)
    }

    # Affichage
    print(f"\n    Entraînement terminé en {training_time:.1f}s")
    print(f"    Résultats TEST :")
    print(f"      • Accuracy : {metrics['Accuracy']:.4f}")
    print(f"      • F1-weighted : {metrics['F1_weighted']:.4f}")
    print(f"      • MCC : {metrics['MCC']:.4f}")
    print(f"      • Balanced Acc : {metrics['Balanced_Accuracy']:.4f}")

    # Libérer mémoire GPU
    del optimizer
    torch.cuda.empty_cache() if torch.cuda.is_available() else None

# ============================================================
# TABLEAU COMPARATIF
# ============================================================
print(f"\n" + "="*70)
print(f" RÉSULTATS FINAUX - MODE TRANSDUCTIF")
print(f"="*70)

import pandas as pd

df_results = pd.DataFrame([
    results_transductive[name]['metrics']
    for name in ['GCN', 'GAT', 'GraphSAGE']
])

# Trier par F1-weighted
df_results = df_results.sort_values('F1_weighted', ascending=False)

print(f"\n{df_results[['Model', 'Accuracy', 'F1_weighted', 'MCC', 'Balanced_Accuracy', 'Training_Time']].to_string(index=False)}")

# Meilleur modèle
best_model_name = df_results.iloc[0]['Model']
best_f1 = df_results.iloc[0]['F1_weighted']

print(f"\n Meilleur modèle : {best_model_name}")
print(f"   F1-weighted : {best_f1:.4f}")

print(f"\n" + "="*70)
print(f" ENTRAÎNEMENT TRANSDUCTIF TERMINÉ")
print(f"="*70)

# Ramener data sur CPU
data = data.to('cpu')
```

```python
# %% Entraînement MODE INDUCTIF (GCN, GAT, GraphSAGE)
import torch
import torch.nn.functional as F
import time
import numpy as np
from torch_geometric.data import Data

print("="*70)
print(" ENTRAÎNEMENT MODE INDUCTIF")
print("="*70)

print(f"\n MODE INDUCTIF :")
print(f"   Le modèle NE VOIT PAS les nœuds de test pendant l'entraînement")
print(f"   → Simulation de l'arrivée de nouveaux nœuds")

# ============================================================
# 1. CRÉER UN SOUS-GRAPHE SANS LES NŒUDS TEST
# ============================================================
print(f"\n🔧 Création du sous-graphe (train + val uniquement)...")

# Masque pour train + val (exclure test)
train_val_mask = data.train_mask | data.val_mask
train_val_indices = torch.where(train_val_mask)[0]

print(f"   Nœuds originaux : {data.num_nodes:,}")
print(f"   Nœuds train+val : {train_val_mask.sum():,}")
print(f"   Nœuds test (exclus) : {data.test_mask.sum():,}")

# Créer un mapping des anciens indices vers les nouveaux
old_to_new = torch.full((data.num_nodes,), -1, dtype=torch.long)
old_to_new[train_val_indices] = torch.arange(train_val_indices.size(0))

# Filtrer les arêtes (garder seulement celles entre nœuds train+val)
edge_mask = train_val_mask[data.edge_index[0]] & train_val_mask[data.edge_index[1]]
edge_index_train_val = data.edge_index[:, edge_mask]

# Remapper les indices des arêtes
edge_index_train_val = old_to_new[edge_index_train_val]

# Créer le nouveau Data object (sous-graphe)
data_inductive_train = Data(
    x=data.x[train_val_mask],
    edge_index=edge_index_train_val,
    y=data.y[train_val_mask]
)

# Créer les nouveaux masques (dans le sous-graphe)
new_train_mask = data.train_mask[train_val_mask]
new_val_mask = data.val_mask[train_val_mask]

data_inductive_train.train_mask = new_train_mask
data_inductive_train.val_mask = new_val_mask

print(f"\n    Sous-graphe créé :")
print(f"      Nœuds : {data_inductive_train.num_nodes:,}")
print(f"      Arêtes : {data_inductive_train.num_edges:,}")
print(f"      Train mask : {new_train_mask.sum():,}")
print(f"      Val mask : {new_val_mask.sum():,}")

# ============================================================
# 2. CRÉER UN GRAPHE POUR LE TEST (train+val+test)
# ============================================================
print(f"\n🔧 Création du graphe complet pour l'inférence test...")

# Pour le test, on utilise le graphe complet mais le modèle
# n'a jamais vu les nœuds test pendant l'entraînement
data_inductive_test = data.clone()

print(f"    Graphe complet (pour inférence) :")
print(f"      Nœuds : {data_inductive_test.num_nodes:,}")
print(f"      Arêtes : {data_inductive_test.num_edges:,}")

# ============================================================
# 3. ENTRAÎNEMENT DES MODÈLES EN MODE INDUCTIF
# ============================================================
device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
data_inductive_train = data_inductive_train.to(device)
data_inductive_test = data_inductive_test.to(device)

models_inductive = {
    'GCN': GCN(num_features, num_classes, hidden_dim=64, dropout=0.5),
    'GAT': GAT(num_features, num_classes, hidden_dim=8, heads=8, dropout=0.6),
    'GraphSAGE': GraphSAGE(num_features, num_classes, hidden_dim=64, dropout=0.5)
}

results_inductive = {}

print(f"\n" + "="*70)

for model_name, model in models_inductive.items():
    print(f"\n🔹 Entraînement {model_name} (MODE INDUCTIF)...")
    print(f"-" * 70)

    model = model.to(device)
    optimizer = torch.optim.Adam(model.parameters(), lr=0.01, weight_decay=5e-4)

    # Early stopping
    best_val_acc = 0
    patience_counter = 0
    best_model_state = None

    # Historique
    history = {
        'train_loss': [],
        'val_loss': [],
        'val_acc': []
    }

    start_time = time.time()

    # Entraînement sur le SOUS-GRAPHE
    for epoch in range(1, 201):
        # Train
        train_loss = train_gnn(model, data_inductive_train, optimizer, device)

        # Evaluate sur train+val (sous-graphe)
        train_acc, _, _, _ = evaluate_gnn(model, data_inductive_train,
                                         data_inductive_train.train_mask, device)
        val_acc, val_loss, _, _ = evaluate_gnn(model, data_inductive_train,
                                              data_inductive_train.val_mask, device)

        # Historique
        history['train_loss'].append(train_loss)
        history['val_loss'].append(val_loss)
        history['val_acc'].append(val_acc)

        # Early stopping
        if val_acc > best_val_acc:
            best_val_acc = val_acc
            patience_counter = 0
            best_model_state = model.state_dict().copy()
        else:
            patience_counter += 1

        # Log
        if epoch % 20 == 0 or epoch == 1:
            print(f"   Epoch {epoch:3d} | Train Loss: {train_loss:.4f} | "
                  f"Val Acc: {val_acc:.4f} | Best Val: {best_val_acc:.4f}")

        # Stop
        if patience_counter >= 30:
            print(f"     Early stopping à l'époque {epoch}")
            break

    # Charger le meilleur modèle
    if best_model_state is not None:
        model.load_state_dict(best_model_state)

    training_time = time.time() - start_time

    # Évaluation sur TEST (nœuds JAMAIS VUS)
    print(f"\n    Évaluation sur nœuds TEST (jamais vus)...")
    test_acc, test_loss, y_true_test, y_pred_test = evaluate_gnn(
        model, data_inductive_test, data_inductive_test.test_mask, device
    )

    # Métriques complètes
    metrics = compute_all_metrics(y_true_test, y_pred_test)
    metrics['Model'] = model_name
    metrics['Training_Time'] = training_time
    metrics['Best_Val_Acc'] = best_val_acc
    metrics['Test_Acc'] = test_acc
    metrics['Mode'] = 'Inductif'

    # Sauvegarder
    results_inductive[model_name] = {
        'model': model,
        'metrics': metrics,
        'history': history,
        'predictions': (y_true_test, y_pred_test)
    }

    # Affichage
    print(f"\n    Entraînement terminé en {training_time:.1f}s")
    print(f"    Résultats TEST (nœuds jamais vus) :")
    print(f"      • Accuracy : {metrics['Accuracy']:.4f}")
    print(f"      • F1-weighted : {metrics['F1_weighted']:.4f}")
    print(f"      • MCC : {metrics['MCC']:.4f}")
    print(f"      • Balanced Acc : {metrics['Balanced_Accuracy']:.4f}")

    # Libérer
    del optimizer
    torch.cuda.empty_cache() if torch.cuda.is_available() else None

# ============================================================
# 4. TABLEAU COMPARATIF INDUCTIF
# ============================================================
print(f"\n" + "="*70)
print(f" RÉSULTATS FINAUX - MODE INDUCTIF")
print(f"="*70)

df_results_inductive = pd.DataFrame([
    results_inductive[name]['metrics']
    for name in ['GCN', 'GAT', 'GraphSAGE']
])

df_results_inductive = df_results_inductive.sort_values('F1_weighted', ascending=False)

print(f"\n{df_results_inductive[['Model', 'Accuracy', 'F1_weighted', 'MCC', 'Balanced_Accuracy', 'Training_Time']].to_string(index=False)}")

# Meilleur modèle inductif
best_inductive = df_results_inductive.iloc[0]['Model']
best_f1_inductive = df_results_inductive.iloc[0]['F1_weighted']

print(f"\n Meilleur modèle (inductif) : {best_inductive}")
print(f"   F1-weighted : {best_f1_inductive:.4f}")

# ============================================================
# 5. COMPARAISON TRANSDUCTIF VS INDUCTIF
# ============================================================
print(f"\n" + "="*70)
print(f"  COMPARAISON TRANSDUCTIF vs INDUCTIF")
print(f"="*70)

# Ajouter la colonne Mode
df_results['Mode'] = 'Transductif'
df_results_inductive['Mode'] = 'Inductif'

# Combiner
df_combined = pd.concat([df_results, df_results_inductive], ignore_index=True)

print(f"\n{'Modèle':<15s} {'Mode':<12s} {'Accuracy':>10s} {'F1-weighted':>12s} {'MCC':>8s}")
print("-"*70)
for _, row in df_combined.iterrows():
    print(f"{row['Model']:<15s} {row['Mode']:<12s} {row['Accuracy']:>10.4f} "
          f"{row['F1_weighted']:>12.4f} {row['MCC']:>8.4f}")

print(f"\n Observations :")
print(f"   • GraphSAGE est conçu pour l'inductif")
print(f"   • GCN/GAT peuvent perdre en performance en mode inductif")
print(f"   • Le mode transductif a accès à plus d'information (tous les nœuds)")

# Ramener sur CPU
data = data.to('cpu')
data_inductive_train = data_inductive_train.to('cpu')
data_inductive_test = data_inductive_test.to('cpu')

print(f"\n" + "="*70)
print(f" ENTRAÎNEMENT INDUCTIF TERMINÉ")
print(f"="*70)
```

```python
import networkx as nx
import matplotlib.pyplot as plt
import numpy as np
from sklearn.manifold import TSNE
import torch
from torch_geometric.utils import to_networkx

# Supposons que vous avez déjà votre objet Data PyTorch Geometric
# data = torch_geometric.data.Data(x, edge_index, y, train_mask, test_mask)

def visualize_graph_structure(data, max_nodes=500):
    """
    Visualise la structure du graphe avec les couleurs des classes
    """
    # Convertir en NetworkX pour visualisation
    G = to_networkx(data, to_undirected=True)

    # Sous-échantillonner si trop de nœuds
    if len(G.nodes()) > max_nodes:
        nodes_to_keep = np.random.choice(list(G.nodes()), max_nodes, replace=False)
        G = G.subgraph(nodes_to_keep)

    # Récupérer les labels des nœuds
    node_colors = [data.y[node].item() for node in G.nodes()]

    # Créer la figure
    fig, ax = plt.subplots(figsize=(15, 10))

    # Layout spring pour mieux espacer les nœuds
    pos = nx.spring_layout(G, k=0.5, iterations=50, seed=42)

    # Dessiner le graphe
    nx.draw_networkx_nodes(
        G, pos,
        node_color=node_colors,
        node_size=50,
        cmap='tab10',
        alpha=0.7,
        ax=ax
    )
    nx.draw_networkx_edges(
        G, pos,
        alpha=0.2,
        width=0.5,
        edge_color='gray',
        ax=ax
    )

    # Légende des classes
    class_names = ['Benign', 'DDOS-HOIC', 'DDOS-LOIC-UDP', 'DDoS-LOIC-HTTP',
                   'DoS-GoldenEye', 'DoS-Hulk', 'DoS-SlowHTTPTest',
                   'DoS-Slowloris', 'FTP-BruteForce', 'SSH-Bruteforce']

    from matplotlib.patches import Patch
    legend_elements = [Patch(facecolor=plt.cm.tab10(i/10), label=class_names[i])
                      for i in range(len(class_names))]
    ax.legend(handles=legend_elements, loc='upper left', fontsize=10)

    plt.title(f'Visualisation du Graphe k-NN ({len(G.nodes())} nœuds, {len(G.edges())} arêtes)',
              fontsize=16, fontweight='bold')
    plt.axis('off')
    plt.tight_layout()
    plt.show()

    print(f"Nombre de composantes connexes : {nx.number_connected_components(G)}")
    print(f"Degré moyen : {sum(dict(G.degree()).values()) / len(G.nodes()):.2f}")
```

```python
def visualize_attention_weights(model, data, node_idx=0, num_hops=2):
    """
    Visualise les poids d'attention du GAT pour un nœud spécifique
    Montre quels voisins reçoivent le plus d'attention
    """
    model.eval()

    # Récupérer les poids d'attention (GAT seulement)
    # Il faut modifier le forward pass du GAT pour retourner les alphas
    with torch.no_grad():
        # Forward pass avec récupération des attentions
        _, attention_weights = model(data.x, data.edge_index, return_attention_weights=True)

    # Extraire le sous-graphe autour du nœud d'intérêt
    edge_index = data.edge_index.cpu().numpy()
    attention = attention_weights[1].cpu().numpy()  # [num_edges]

    # Trouver les arêtes connectées au nœud
    ego_edges = np.where((edge_index[0] == node_idx) | (edge_index[1] == node_idx))[0]

    # Créer un sous-graphe
    G = nx.Graph()
    for idx in ego_edges:
        src, dst = edge_index[0, idx], edge_index[1, idx]
        weight = attention[idx]
        G.add_edge(src, dst, weight=weight)

    # Visualisation
    fig, ax = plt.subplots(figsize=(12, 8))
    pos = nx.spring_layout(G, seed=42)

    # Nodes
    node_colors = [data.y[node].item() for node in G.nodes()]
    nx.draw_networkx_nodes(G, pos, node_color=node_colors, node_size=300,
                           cmap='tab10', alpha=0.8, ax=ax)

    # Edges avec épaisseur proportionnelle à l'attention
    edges = G.edges()
    weights = [G[u][v]['weight'] for u, v in edges]
    nx.draw_networkx_edges(G, pos, width=np.array(weights)*5,
                           alpha=0.6, edge_color=weights,
                           edge_cmap=plt.cm.Reds, ax=ax)

    # Labels
    nx.draw_networkx_labels(G, pos, font_size=10, ax=ax)

    plt.title(f'Attention Weights GAT - Nœud {node_idx} (classe: {data.y[node_idx].item()})',
              fontsize=14, fontweight='bold')
    plt.axis('off')
    plt.colorbar(plt.cm.ScalarMappable(cmap=plt.cm.Reds),
                 ax=ax, label='Attention Weight')
    plt.tight_layout()
    plt.show()
```

```python
def plot_training_curves(train_losses, val_losses, train_accs, val_accs):
    """
    Visualise l'évolution de la loss et accuracy pendant l'entraînement
    """
    fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(16, 5))

    epochs = range(1, len(train_losses) + 1)

    # Loss
    ax1.plot(epochs, train_losses, 'b-', label='Train Loss', linewidth=2)
    ax1.plot(epochs, val_losses, 'r-', label='Val Loss', linewidth=2)
    ax1.set_xlabel('Epoch', fontsize=12)
    ax1.set_ylabel('Loss', fontsize=12)
    ax1.set_title('Evolution de la Loss', fontsize=14, fontweight='bold')
    ax1.legend(fontsize=11)
    ax1.grid(True, alpha=0.3)

    # Accuracy
    ax2.plot(epochs, train_accs, 'b-', label='Train Accuracy', linewidth=2)
    ax2.plot(epochs, val_accs, 'r-', label='Val Accuracy', linewidth=2)
    ax2.set_xlabel('Epoch', fontsize=12)
    ax2.set_ylabel('Accuracy (%)', fontsize=12)
    ax2.set_title('Evolution de l\'Accuracy', fontsize=14, fontweight='bold')
    ax2.legend(fontsize=11)
    ax2.grid(True, alpha=0.3)
    ax2.set_ylim([0, 100])

    plt.tight_layout()
    plt.show()
```

```python
def visualize_message_passing(data, node_idx=0, num_layers=2):
    """
    Visualise comment l'information se propage dans le graphe
    à travers les couches du GNN
    """
    # Convertir en NetworkX
    G = to_networkx(data, to_undirected=True)

    fig, axes = plt.subplots(1, num_layers + 1, figsize=(18, 5))

    # Calculer les voisinages à différentes profondeurs
    for layer in range(num_layers + 1):
        ax = axes[layer]

        if layer == 0:
            # Nœud initial
            nodes_at_layer = {node_idx}
            title = f'Layer 0: Nœud Initial'
        else:
            # Voisins à distance layer
            nodes_at_layer = set(nx.single_source_shortest_path_length(
                G, node_idx, cutoff=layer).keys())
            title = f'Layer {layer}: {len(nodes_at_layer)} nœuds'

        # Sous-graphe
        subG = G.subgraph(nodes_at_layer)
        pos = nx.spring_layout(subG, seed=42)

        # Couleurs selon la distance au nœud central
        distances = nx.single_source_shortest_path_length(G, node_idx, cutoff=layer)
        node_colors = [distances.get(n, layer) for n in subG.nodes()]

        # Dessiner
        nx.draw_networkx_nodes(subG, pos, node_color=node_colors,
                              node_size=100, cmap='YlOrRd',
                              vmin=0, vmax=num_layers, ax=ax)
        nx.draw_networkx_edges(subG, pos, alpha=0.3, ax=ax)

        # Mettre en évidence le nœud central
        if node_idx in subG:
            nx.draw_networkx_nodes(subG, pos, nodelist=[node_idx],
                                  node_color='red', node_size=200,
                                  node_shape='*', ax=ax)

        ax.set_title(title, fontsize=12, fontweight='bold')
        ax.axis('off')

    plt.suptitle('Propagation du Message dans le GNN',
                 fontsize=16, fontweight='bold', y=1.02)
    plt.tight_layout()
    plt.show()
```

```python
def plot_per_class_performance(y_true, y_pred):
    """
    Visualise les performances du GNN par classe avec confusion matrix
    """
    from sklearn.metrics import confusion_matrix, classification_report
    import seaborn as sns

    class_names = ['Benign', 'DDOS-HOIC', 'DDOS-LOIC-UDP', 'DDoS-LOIC-HTTP',
                   'DoS-GoldenEye', 'DoS-Hulk', 'DoS-SlowHTTPTest',
                   'DoS-Slowloris', 'FTP-BruteForce', 'SSH-Bruteforce']

    # Confusion Matrix
    cm = confusion_matrix(y_true, y_pred)

    fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(18, 7))

    # Heatmap de la matrice de confusion
    sns.heatmap(cm, annot=True, fmt='d', cmap='Blues',
                xticklabels=class_names, yticklabels=class_names,
                cbar_kws={'label': 'Nombre de prédictions'},
                ax=ax1)
    ax1.set_xlabel('Classe Prédite', fontsize=12)
    ax1.set_ylabel('Classe Réelle', fontsize=12)
    ax1.set_title('Matrice de Confusion GNN', fontsize=14, fontweight='bold')
    plt.setp(ax1.get_xticklabels(), rotation=45, ha='right')
    plt.setp(ax1.get_yticklabels(), rotation=0)

    # F1-Score par classe
    from sklearn.metrics import f1_score
    f1_scores = []
    for i in range(len(class_names)):
        mask_true = (y_true == i)
        mask_pred = (y_pred == i)
        if mask_true.sum() > 0:
            f1 = f1_score(mask_true, mask_pred)
            f1_scores.append(f1)
        else:
            f1_scores.append(0)

    colors = plt.cm.RdYlGn(np.array(f1_scores))
    bars = ax2.barh(class_names, f1_scores, color=colors)
    ax2.set_xlabel('F1-Score', fontsize=12)
    ax2.set_title('F1-Score par Classe d\'Attaque', fontsize=14, fontweight='bold')
    ax2.set_xlim([0, 1])
    ax2.grid(axis='x', alpha=0.3)

    # Ajouter les valeurs sur les barres
    for i, (bar, score) in enumerate(zip(bars, f1_scores)):
        ax2.text(score + 0.02, i, f'{score:.3f}',
                va='center', fontsize=10, fontweight='bold')

    plt.tight_layout()
    plt.show()
```
